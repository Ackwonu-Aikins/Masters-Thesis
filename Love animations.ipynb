```python
"""
Particle-assembly crucifix, v2.

The previous version only formed a hatched "cloud" in roughly the right
place, which reads as texture, not a figure. This version keeps that
particle-assembly effect for atmosphere/shading, then draws a crisp,
continuous line-art figure of Christ on top: face with closed eyes and
beard, crown of thorns, ribbed torso, arms tapering to hands with
fingers, a draped loincloth with a hanging fold, and bent, crossed legs
ending in feet with a nail mark -- so the body is clearly recognizable,
not just a scatter of dashes.
"""

import turtle as t
import math as m
import random as r

# ----------------------------------------------------------------------
# Screen setup
# ----------------------------------------------------------------------
s = t.Screen()
s.setup(700, 800)
s.bgcolor("white")
s.tracer(0)

INK = "#0a0a0a"


def lerp(a, b, u):
    return a + (b - a) * u


# ========================================================================
# PART 1 — background particle hatch (wood grain + soft body shading)
# ========================================================================
def region_points(center_fn, width_fn, t0, t1, n):
    pts = []
    dt = 1e-3
    for _ in range(n):
        tt = r.uniform(t0, t1)
        cx, cy = center_fn(tt)
        cx2, cy2 = center_fn(min(tt + dt, t1))
        ang = m.atan2(cy2 - cy, cx2 - cx)
        perp = ang + m.pi / 2
        w = width_fn(tt)
        off = r.uniform(-w / 2, w / 2)
        x = cx + m.cos(perp) * off
        y = cy + m.sin(perp) * off
        pts.append((x, y, m.degrees(ang)))
    return pts


def vbeam_center(tt):
    return (0, lerp(-330, 300, tt))


def vbeam_width(tt):
    if tt > 0.93:
        return lerp(60, 4, (tt - 0.93) / 0.07)
    return 60


def hbeam_center(tt):
    return (lerp(-235, 235, tt), 165)


def hbeam_width(tt):
    edge = min(tt, 1 - tt)
    taper = min(1.0, edge / 0.08)
    return lerp(30, 70, taper)


def torso_shade_center(tt):
    return (lerp(0, -4, tt), lerp(150, 45, tt))


def torso_shade_width(tt):
    return lerp(70, 30, tt) - m.sin(tt * m.pi) * 6


def cloth_shade_center(tt):
    x = lerp(-6, -35, tt) + m.sin(tt * m.pi * 1.3) * 10
    y = lerp(42, -60, tt)
    return (x, y)


def cloth_shade_width(tt):
    return lerp(55, 20, tt)


def legs_shade_center(tt):
    x = lerp(0, 8, tt) + m.sin(tt * m.pi) * -8
    y = lerp(-65, -300, tt)
    return (x, y)


def legs_shade_width(tt):
    return lerp(40, 16, tt)


targets = []
targets += [(*p2, "wood", r.uniform(7, 13)) for p2 in region_points(vbeam_center, vbeam_width, 0, 1, 420)]
targets += [(*p2, "wood", r.uniform(7, 13)) for p2 in region_points(hbeam_center, hbeam_width, 0, 1, 320)]
targets += [(*p2, "flesh", r.uniform(4, 8)) for p2 in region_points(torso_shade_center, torso_shade_width, 0, 1, 140)]
targets += [(*p2, "cloth", r.uniform(4, 8)) for p2 in region_points(cloth_shade_center, cloth_shade_width, 0, 1, 100)]
targets += [(*p2, "flesh", r.uniform(4, 8)) for p2 in region_points(legs_shade_center, legs_shade_width, 0, 1, 130)]

particles = []
for tx, ty, ang, kind, length in targets:
    sx, sy = r.randint(-380, 380), r.randint(-380, 380)
    particles.append([sx, sy, tx, ty, ang, length])

p = t.Turtle()
p.hideturtle()
p.pencolor(INK)


def draw_stroke(x, y, ang, length, width=1.0):
    p.width(width)
    rad = m.radians(ang)
    dx, dy = m.cos(rad) * length / 2, m.sin(rad) * length / 2
    p.penup()
    p.goto(x - dx, y - dy)
    p.pendown()
    p.goto(x + dx, y + dy)


# ========================================================================
# PART 2 — the crisp line-art figure (drawn once assembly settles)
# ========================================================================
def draw_path(pts, width=2.2, close=False):
    p.width(width)
    p.penup()
    p.goto(pts[0])
    p.pendown()
    for pt in pts[1:]:
        p.goto(pt)
    if close:
        p.goto(pts[0])


def draw_ribbon(upper, lower, width=2.0):
    draw_path(upper, width)
    draw_path(list(reversed(lower)), width)
    p.penup()
    p.goto(upper[-1])
    p.pendown()
    p.goto(lower[-1])
    p.penup()
    p.goto(upper[0])
    p.pendown()
    p.goto(lower[0])


def draw_figure():
    # --- head ---
    head = [(-6, 222), (10, 224), (20, 215), (24, 200), (20, 182),
            (10, 168), (0, 163), (-10, 168), (-18, 180), (-20, 198),
            (-14, 214), (-6, 222)]
    draw_path(head, 2.2)

    # crown of thorns (jagged ring)
    crown = []
    cx, cy = 2, 196
    spikes = 22
    for i in range(spikes + 1):
        ang = i / spikes * 2 * m.pi
        rad = 30 if i % 2 == 0 else 23
        crown.append((cx + m.cos(ang) * rad, cy + m.sin(ang) * rad))
    draw_path(crown, 1.4)

    # closed eyes, brow, nose, mouth
    draw_path([(-10, 202), (-3, 200)], 1.6)
    draw_path([(5, 200), (12, 202)], 1.6)
    draw_path([(1, 198), (-1, 188), (2, 185)], 1.4)
    draw_path([(-4, 178), (4, 178)], 1.4)

    # beard, hanging down over the chest
    beard = [(-14, 178), (-16, 165), (-10, 150), (-4, 145), (2, 150),
             (8, 148), (4, 158), (0, 168), (4, 176), (-2, 180), (-14, 178)]
    draw_path(beard, 1.6)

    # --- shoulders / collarbone ---
    draw_path([(-40, 148), (-20, 153), (0, 150), (20, 153), (40, 148)], 2.0)

    # --- arms, tapering to hands ---
    l_upper = [(-38, 150), (-90, 158), (-140, 164), (-183, 169)]
    l_lower = [(-38, 130), (-95, 140), (-145, 149), (-181, 158)]
    draw_ribbon(l_upper, l_lower, 2.0)
    for fx, fy in [(-190, 175), (-195, 168), (-197, 160), (-193, 152)]:
        draw_path([(-183, 163), (fx, fy)], 1.3)

    r_upper = [(38, 150), (90, 158), (140, 164), (183, 169)]
    r_lower = [(38, 130), (95, 140), (145, 149), (181, 158)]
    draw_ribbon(r_upper, r_lower, 2.0)
    for fx, fy in [(190, 175), (195, 168), (197, 160), (193, 152)]:
        draw_path([(183, 163), (fx, fy)], 1.3)

    # --- torso outline + ribs + sternum ---
    torso_l = [(-40, 148), (-42, 118), (-34, 90), (-24, 60), (-16, 40)]
    torso_r = [(40, 148), (42, 118), (34, 90), (24, 60), (16, 40)]
    draw_path(torso_l, 2.0)
    draw_path(torso_r, 2.0)
    draw_path([(0, 148), (0, 45)], 1.4)
    for ry in (132, 116, 100, 84):
        span = lerp(30, 18, (132 - ry) / 48)
        draw_path([(-span, ry), (0, ry - 4), (span, ry)], 1.1)

    # --- loincloth, draped left with a hanging fold ---
    cloth = [(-16, 40), (-45, 12), (-55, -18), (-46, -50), (-24, -68),
             (0, -62), (14, -45), (20, -20), (16, 40)]
    draw_path(cloth, 2.2, close=True)
    fold = [(-38, -10), (-42, -45), (-32, -80), (-18, -105), (-10, -95)]
    draw_path(fold, 1.8)
    for fy0, fy1 in [(-5, -35), (-15, -50), (-28, -62)]:
        draw_path([(-30, fy0), (-24, fy1)], 1.0)
    # small gathered knot at the waist
    draw_path([(-6, 38), (-2, 30), (4, 36), (0, 42), (-6, 38)], 1.3)

    # --- legs, bent and crossed, to a nailed foot ---
    left_leg_u = [(-20, -60), (-15, -140), (-8, -215), (-2, -280)]
    left_leg_l = [(-2, -60), (2, -140), (6, -210), (10, -278)]
    draw_ribbon(left_leg_u, left_leg_l, 2.0)

    right_leg_u = [(16, -60), (24, -130), (16, -205), (8, -270)]
    right_leg_l = [(2, -60), (10, -130), (4, -205), (2, -270)]
    draw_ribbon(right_leg_u, right_leg_l, 2.0)

    # crossed feet + toes + nail
    feet = [(-4, -282), (2, -300), (12, -308), (20, -302), (14, -288),
            (6, -278), (-4, -282)]
    draw_path(feet, 2.0, close=True)
    for tx2, ty2 in [(18, -312), (22, -306), (24, -298)]:
        draw_path([(16, -302), (tx2, ty2)], 1.1)
    nail = [(6, -298), (12, -292)]
    draw_path(nail, 1.6)
    nail2 = [(9, -301), (9, -289)]
    draw_path(nail2, 1.6)


# ----------------------------------------------------------------------
# Animate the background hatch in, then reveal the figure on top
# ----------------------------------------------------------------------
STEPS = 60
for frame in range(STEPS + 1):
    p.clear()
    u = frame / STEPS
    eased = m.sin(u * m.pi / 2)

    for pt in particles:
        sx, sy, tx, ty, ang, length = pt
        cx = lerp(sx, tx, eased)
        cy = lerp(sy, ty, eased)
        cur_len = lerp(length * 0.4, length, eased)
        cur_ang = lerp(r.uniform(0, 360) if frame == 0 else ang, ang, eased)
        draw_stroke(cx, cy, cur_ang, cur_len, width=1.0)

    if frame == STEPS:
        draw_figure()

    s.update()

t.done()
```
