```python
%matplotlib
import matplotlib.pyplot as plt
import numpy as np
plt.style.use('seaborn-v0_8-whitegrid')
```

```python
#A histogram
rng = np.random.default_rng(seed=50)
data = rng.standard_normal(1000)
plt.hist(data)
```

```python
#A customized histogram
rng = np.random.default_rng(seed=42)
data = rng.standard_normal(1000)
plt.hist(data, bins = 30, histtype = 'stepfilled',alpha = 0.5, color = "green")
```

```python
#Over- plotting multiple histograms
x1 = rng.normal(0,0.8,1000)
x2 = rng.normal(-2,1,1000)
x3 = rng.normal(3,2,1000)

kwargs = dict(histtype = 'stepfilled', alpha= 0.4, bins = 40)
plt.hist(x1, **kwargs)
plt.hist(x2, **kwargs)
plt.hist(x3, **kwargs)
```

```python
counts, bin_edges = np.histogram(x2, bins = 40) #np.histogram computes the histogram of the data given
counts #the number of points that fall inside a bin
```

```python
bin_edges #the bins in which the counts can be found
```

```python
mean = [0,0]
cov = [[1,1],[1,2]]
x, y = rng.multivariate_normal(mean,cov,10000).T
plt.hist2d(x,y, bins = 30, cmap ="Blues")
cb = plt.colorbar()
cb.set_label('counts in bin')
```
