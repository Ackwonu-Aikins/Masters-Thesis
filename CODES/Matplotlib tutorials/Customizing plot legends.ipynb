```python
import matplotlib.pyplot as plt
plt.style.use('seaborn-v0_8-whitegrid')
import numpy as np
```

```python
x = np.linspace(0,10,1000)
fig, ax = plt.subplots()
ax.plot(x,np.sin(x), '-b', label = 'Sine')
ax.plot(x,np.cos(x), '--r', label = 'Cosine')
ax.axis('equal')
leg = ax.legend(loc = 'lower left', ncol = 2)
```
