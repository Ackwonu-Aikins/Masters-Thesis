```python
import matplotlib.pyplot as plt
plt.style.use('seaborn-v0_8-white')
import numpy as np
```

```python
#Example of an inset axes
ax1 = plt.axes()
ax1 = plt.axes([0.5,0.5,0.3,0.3])
```

```python
fig = plt.figure()
ax1 = fig.add_axes([0.1, 0.5, 0.8, 0.4],xticklabels=[], ylim=(-1.2, 1.2))
ax2 = fig.add_axes([0.1, 0.1, 0.8, 0.4],ylim=(-1.2, 1.2))
x = np.linspace(0,10)
ax1.plot(np.sin(x))
ax2.plot(np.cos(x))
```

```python
fig, ax = plt.subplots(1,2) # rows , columns
x = np.linspace(0,10)
ax[0].plot(x**2)
ax[1].plot([1,2,3,4],[1,2,3,6])

ax[0].set( xlabel = 'x', ylabel = 'sin(x)', title = 'Quadratic')
ax[1].set( xlabel = 'x', ylabel = 'f(x)', title = 'Simple graph')
```

```python
fig, ax = plt.subplots(2,2,  figsize = (4,4))
ax[0,0].plot(x**3)
ax[0,1].plot(np.cos(x))
ax[1,0].plot(np.sin(x))
ax[1,1].plot(np.tan(x))
```

```python
rng = np.random.default_rng(seed = 42)
x = rng.normal(0,1,100)
y = rng.normal(0,1,100)

fig, ax = plt.subplots(1,2)
ax[0].hist(x, bins = 30,histtype = 'bar',label = 'bar')
ax[0].set(xlabel = 'x', ylabel = 'f(x)',title = 'A bar chart')
ax[0].legend()
ax[1].scatter(x,y,marker = '+',c = np.sin(x), cmap = 'viridis',alpha = 0.5)
ax[1].set(xlabel = 'x', ylabel = 'f(x)', title = 'Scatter Plot')
```

```python
fig, axs = plt.subplots(2, 2,sharex=True)

for ax in axs.flat:
    ax.plot([1,2,3], [1,4,9])
```
