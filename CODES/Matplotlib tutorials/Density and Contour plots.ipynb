```python
%matplotlib
import matplotlib.pyplot as plt
plt.style.use('seaborn-v0_8-whitegrid')
import numpy as np
```

```python
#Function "f" with 2 inputs x and y
def f(x,y):
    return np.sin(x) ** 10 + np.cos(10+y*x) * np.cos(x)
```

```python
#Visualizing 3D data with contours
x = np.linspace(0,5,50)
y = np.linspace(0,5,40)
X,Y = np.meshgrid(x,y)

Z = f(X,Y); #function "f" was called to evaluate the Z coordinates
#plt.contour(X,Y,Z,cmap = 'Accent')
#plt.cm.<tab> for more colormaps(cmaps)

'''The Lines (plt.contour): These are called isolines. Every point on a single line has the exact same 
value. If the lines are very close together, the "hill" is very steep. If they are far apart, the terrain is flat.
The Colors (plt.contourf): The "f" stands for filled. This fills the space between the lines with color gradients.
Usually, darker or cooler colors (blues) represent "valleys" (low 
), and brighter or warmer colors (yellows/reds) represent "peaks" (high).'''
```

```python
#Visualizing with filled contour (plt.contourf)
plt.contourf(X,Y,Z, levels = 4, cmap = 'RdGy')
plt.colorbar()
```

```python
plt.imshow(Z, extent=[0, 5, 0, 5], origin='upper',cmap='RdGy')
plt.colorbar()
plt.axis('image')
```
