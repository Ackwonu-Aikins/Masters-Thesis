```python
import matplotlib.pyplot as plt
plt.style.use('seaborn-v0_8-whitegrid')
import numpy as np
```

```python
fig = plt.figure()
ax = plt.axes()

x = np.linspace(0,10,100)
ax.plot(x, np.sin(x),'r:');
#ax.plot(x,np.cos(x))
```

```python
plt.plot(x, np.sin(x))
```

```python
#Multiple lines in one figure using plt
plt.plot(x, np.sin(x))
plt.plot(x, np.cos(x));
```

```python
#Coloring in plt.plot
plt.plot(x, np.sin(x -0), color = 'g')
```

```python
#Linestyle 
plt.plot(x, x +0 , linestyle = 'solid')
plt.plot(x, x+1, linestyle = 'dashed')
plt.plot(x, x+2, linestyle = ':')#dotted
plt.plot(x, x+3, linestyle = '-.')#dashdot
plt.plot(x, x+4, '.g')#combined color and linestyle format
```

```python
#Setting limit of the plot
plt.plot(x, np.sin(x))

plt.xlim(-1,11)
plt.ylim(-1.5, 1.5);
```

```python
#Setting the axis using plt.axis method
plt.plot(x, np.sin(x))
plt.axis([-1,11,-1.5,1.5])
```

```python
#Optional values that the plt.axis can take
plt.plot(x, np.sin(x))
plt.axis('tight')#1
plt.axis('equal')#2 
#more dey
```

```python
#Adding label, title and legends
plt.plot(x,np.sin(x))
plt.title("A Sine Curve") #title
plt.xlabel('x') #x axis label
plt.ylabel('y') #y axis label
```

```python
plt.plot(x, np.sin(x), '-k', label = 'sin(x)')
plt.plot(x, np.cos(x), '-.r', label = 'cos(x)')
plt.axis('equal')
plt.legend();
```

```python
ax = plt.axes()
ax.plot(x, np.sin(x), 'r>-')
ax.set(xlim = (0,10), ylim = (-2,2), xlabel = 'x', ylabel ='sin(x)',
       title = 'A Simple plot');
```
