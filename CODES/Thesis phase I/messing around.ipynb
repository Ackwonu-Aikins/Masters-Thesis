```python
import matplotlib.pyplot as plt
```

```python
#num_vehicle = 2
routes = {
    'customer':[0,1,2,3,4],
    'nodes': [('0--> 3--> 0'),('0--> 7--> 0')],
    'distance': [200,300],
    'load': ['400m','400m '],
    'coordinates':[[(4,5)],
                    [(7,3)],
                    [(4,2)],
                    [(2,3)],
                    [(8,2)]]}

fig, ax = plt.subplots(1,figsize = (9,8))
for r in range(len(routes['customer'])):
    coord = routes['coordinates'][r]
    for i in range(len(coord)):
        x,y = coord[i]
        ax.scatter(x,y)
        ax.text(x, y,s = r,ha = 'center', va = 'center',fontsize=15)
```
